# Image to PDF Converter

This notebook converts image files from a selected folder into PDF documents using the `img2pdf` tool. It scans the input directory for common image formats, creates a PDF for each image, and saves the results in an output folder.

## What you need
- A folder containing image files such as `.jpg`, `.jpeg`, `.png`, `.tif`, or `.tiff`
- The `img2pdf` command installed in your environment
- An input folder path that you want to convert


```python
# Import required libraries
import os
import subprocess
from pathlib import Path

# The notebook uses these modules to interact with the filesystem and run the external conversion tool.
```

```python
# Configure input and output folders
input_folder = Path('/Users/cristofer/Documents/99 - Personal Stuff/Docs Pessoais')
output_folder = input_folder / "pdf_output"
dpi = 150  # adjust between 100–300
page_size = "A4"  # can be 'A4', 'letter', or custom like '210mmx297mm'

# Create output folder if it doesn't exist
output_folder.mkdir(exist_ok=True)

# Supported image extensions for conversion
image_extensions = [".jpg", ".jpeg", ".png", ".tif", ".tiff"]

# Update these values if you want to use a different source folder or output location.
```

```python
# Convert each supported image into a PDF
for file in sorted(input_folder.iterdir()):
    if file.suffix.lower() in image_extensions:
        output_pdf = output_folder / f"{file.stem}.pdf"

        # Build the img2pdf command
        cmd = [
            "img2pdf",
            str(file),
            "--pagesize", page_size,
            "--fit", "into",
            "--auto-orient",
            "-o", str(output_pdf)
        ]

        print(f"Converting: {file.name} → {output_pdf.name}")
        subprocess.run(cmd, check=True)

print("\n✅ All images converted successfully!")
print(f"PDFs saved in: {output_folder}")
```
